```python

5.Create a project using logistic regression with tensor?

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
score = tf.constant([20, 35, 12, 65, 75, 90], dtype=tf.float32)
output = tf.constant([0, 0, 0, 1, 1, 1], dtype=tf.float32)

m_eager = tf.Variable(0.0)
b_eager = tf.Variable(0.0)

learning_rate = 0.001
epochs = 100

print("Training using Eager Execution")

for i in range(epochs):
    with tf.GradientTape() as tape:
        logits = m_eager * score + b_eager
        loss = tf.reduce_mean(
            tf.nn.sigmoid_cross_entropy_with_logits(
                labels=output, logits=logits
            )
        )

    grads = tape.gradient(loss, [m_eager, b_eager])
    m_eager.assign_sub(learning_rate * grads[0])
    b_eager.assign_sub(learning_rate * grads[1])

    if i % 20 == 0:
        print(f"[Eager] Step {i} : Loss = {loss.numpy():.4f}")

m_graph = tf.Variable(0.0)
b_graph = tf.Variable(0.0)

@tf.function
def train_step(x, y, m, b):
    with tf.GradientTape() as tape:
        logits = m * x + b
        y_pred = tf.nn.sigmoid(logits)
        loss = tf.reduce_mean(
            tf.keras.losses.binary_crossentropy(y, y_pred)
        )

    grads = tape.gradient(loss, [m, b])
    m.assign_sub(learning_rate * grads[0])
    b.assign_sub(learning_rate * grads[1])

    return loss


print("\nTraining using Graph Execution")

for i in range(epochs):
    loss = train_step(score, output, m_graph, b_graph)

    if i % 20 == 0:
        print(f"[Graph] Step {i} : Loss = {loss.numpy():.4f}")

score_range = np.linspace(
    float(tf.reduce_min(score)),
    float(tf.reduce_max(score)),
    100
)

plt.scatter(score, output, label="True Data", color='black')

plt.plot(
    score_range,
    tf.nn.sigmoid(m_eager * score_range + b_eager),
    label="Eager Model",
    linestyle="--"
)

plt.plot(
    score_range,
    tf.nn.sigmoid(m_graph * score_range + b_graph),
    label="Graph Model"
)

plt.xlabel("Marks")
plt.ylabel("Probability")
plt.title("Logistic Regression using TensorFlow")
plt.legend()
plt.grid(True)
plt.show()
```

1. Explain the role of weights in a neuron

Weights are numerical values assigned to each input of a neuron in a neural network. They determine the importance or influence of each input feature on the output. During training, the model adjusts the weights to minimize error and improve prediction accuracy. Higher weight means the input has more impact on the final output.

2. What is an activation function?

An activation function is a mathematical function applied to the output of a neuron to introduce non-linearity into the model. It helps the neural network learn complex patterns. Common activation functions include: Sigmoid,ReLU,Tanh,Softmax

3. Define probability distribution in ML context

A probability distribution describes how the probabilities are assigned to different possible outcomes of a random variable. In machine learning, probability distributions are used to model uncertainty, make predictions, and estimate parameters. Examples include:Normal distribution,Bernoulli distribution,Gaussian distribution

4. What is gradient in optimization?

A gradient is a vector that contains the partial derivatives of a function with respect to its parameters. In machine learning optimization, gradients indicate the direction and rate of fastest increase of the loss function. Algorithms like gradient descent use gradients to update model parameters to minimize error.
